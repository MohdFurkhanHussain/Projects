```python
# Broadcasting with arrays of different shapes
```

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([10, 20, 30])

result = arr1 + arr2
result
```

```python
arr1 = np.array([[1, 2, 3], [4, 5, 6]])   # 2D
arr2 = np.array([1, 2, 3])                # 1D

result = arr1 + arr2                      # 2D + 1D
result
```

```python
# Simulating a dataset (5 sample, 3 feature)

data = np.array([[10, 20, 30],
                [15, 25, 35],
                [20, 30, 40],
                [25, 35, 45],
                [30, 40 ,50]])

# Calcuating the mean and standard divisation for each feature.
mean = data.mean(axis = 0)
std = data.std(axis = 0)

# Normalisation the data using Broadcasting
normalized_data = (data - mean) / std

print(normalized_data)
```

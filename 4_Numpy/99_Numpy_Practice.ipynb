## Project 2: Intermediate Data Analytics – Word Frequency Counter
Using Numpy

Analyze any block of text to show how often each word appears.

```python
import numpy as np

text = ("Hey I am Mohd Furkhan Hussain and I am practicing python, numpy to get ahead in life.")
text = text.lower()
words = text.split()

words_array = np.array(words)
unique_words, word_counts = np.unique(words_array, return_counts=True)

print("word frequency:")
for word, count in zip(unique_words, word_counts):
    print(f"{word} : {count}")
print(len(text))
```

### Using a Short method

```python
import numpy as np
words = ("This is the shortest method, this is amazing").lower().split()
w, c = np.unique(words, return_counts=True)
for word, count in zip(w, c):
    print(word, ":", count)
```

```python
import numpy as np

portfolio = [
    {"name": "Apple", "shares": 2, "price": 145},
    {"name": "Google", "shares": 3, "price": 2750},
    {"name": "Infosys", "shares": 5, "price": 1620}
]

# Make lists for shares and prices
shares = np.array([stock["shares"] for stock in portfolio])
prices = np.array([stock["price"] for stock in portfolio])

# Multiply shares and prices, then add up everything
values = shares * prices

# Print each company's value
for i, stock in enumerate(portfolio):
    print(stock["name"], ":", values[i])

# Print total portfolio value
print("Total portfolio value:", np.sum(values))
```

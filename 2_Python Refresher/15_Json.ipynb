### json module
dump, dumps, load, loads.

```python
import json
```

```python
obj = ({"name": "Mohd Furkhan Hussain", "company": "Khandanii", "Networth": "75,000,000,000.00 $"})
json.dumps(obj)
```

```python
obj = ({"name": "Mohd Furkhan Hussain", "company": "Khandanii", "Networth": "75,000,000,000.00 $"})
s = json.dumps(obj)
print(s)
```

```python
d = json.loads('{"name": "Mohd Furkhan Hussain", "company": "Khandanii", "Networth": "75,000,000,000.00 $"}')
```

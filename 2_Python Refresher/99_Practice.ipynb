### Class representing a dataset (Simple model)

```python
class dataset:
    def __init__(self, data):
        self.data = data

    def add_row(self, row):
        self.data.append(row)

    def num_rows(self):
        return len(self.data)     # Till here is the main formula

# Example Usage
dataset = dataset([[1,2], [3,4]])
dataset.add_row([5,6])
print(dataset.num_rows())
```

### Class Inheritance: Shape area/parimeter

```python
import math

class shape:
    def area(self):
        pass

    def parimeter(self):
        pass

class circle (shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius

# Example
circle = circle(5)
print(circle.area())
print(circle.perimeter())
```

```python
#word analysis

txt = input("I am gong to practice PTE for couple of days. And I am going to learn and practice Data scientist till I'll be prefect in it: ")
fixed_txt = txt.lower()
words = fixed_txt.split()
freq = {}

for word in words:
    if word in freq:
        freq[word] += +1

    else:
        freq[word] = 1

for word, count in freq.items():
    print(f'{word}: {count}')
```

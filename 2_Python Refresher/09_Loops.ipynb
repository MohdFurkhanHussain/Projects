```python
l = ["Apples", "Bananas", "Watermelon"]
for item in l:
    if (item == "Bananas"):
        break      # Stops thi loop right here
    print(item)
    
```

```python
l = ["Apples", "Bananas", "Watermelon"]
for item in l:
    if (item == "Bananas"):
        continue     # Skips this particular iteraction of the loop
    print(item)
    
```

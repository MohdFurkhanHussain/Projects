### Match case

```python
code = 500
```

```python
match code:
    case 200:
        print("response ok")
    case 400:
        print("Bad response")
    case 404:
        print("Not Found")
print ("End of program")
```

```python
code = 200
```

```python
match code:
    case 200:
        print("response ok")
        print("Good Response")
    case 400:
        print("Bad response")
    case 404:
        print("Not Found")
print ("End of program")
```

```python
code = 2000
```

```python
match code:
    case 200:
        print("response ok")
    case 400:
        print("Bad response")
    case 404:
        print("Not Found")
    case _:
        print("Nothinng Matched")
print ("End of program")
```

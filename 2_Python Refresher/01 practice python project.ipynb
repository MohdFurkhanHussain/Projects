1. Calculate Simple Moving Average Manually
   
Why use this?
A moving average smooths out fluctuations in time series data, helping spot trends and remove noise. Use this as a fast baseline forecast for sales or stock prices when datasets are small or tools are limited.

How do we use it?
When you have a list of values (like monthly sales) and want to forecast or smooth them, average each consecutive window of values.

When?
Useful for short-term trend forecasting and for learning how moving averages work step-by-step.



```python
# Sales data for six months

sales = [120, 140, 135, 150, 160, 170]
window = 3

moving_average = []
for i in range(len(sales) - window + 1):
    s = 0
    for j in range(window):
        s+= sales[i + j] # sum the next 'window' value
    avg = s/window # Calcuate the mean
    moving_average.append(avg) # Store the result

print("original sales", sales)
print("3-point moving average", moving_average)
```

```python
sales = [100000000, 200000000, 400000000, 750000000, 850000000, 1000000000]
months = 3

moving_average = []
for i in range(len(sales)- months + 1):
    s = 0
    for j in range(months):
        s += sales[i + j]
    avg = s/months
    moving_average.append(avg)

print("Original sales of KHANDANII", sales)
print("3-point moving average", moving_average)
```

```python
#3. Find Outlier Points in Finance Data

expenses = [200, 210, 215, 205, 800, 220, 210] # one value is much higher
n = len(expenses)
mean = sum(expenses)/n

# Compute differences and thresholdfor outlier (e.g 2 times average from mean)
differences = [abs(x - mean) for x in expenses]
threshold = 2 * (sum(differences)/n)

outliers = []
for i in range (n):
    if abs(expenses[i] - mean) > threshold:
        outliers.append(expenses[i])

print("expenses:", expenses )
print("mean:", mean)
print("outliers", outliers)
```

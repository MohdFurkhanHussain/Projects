### Project 1: Basic Data Science – Average Calculator

**Goal:**  
Take a series of numbers (like exam scores) from the user, calculate the average, and understand basic statistics.

```python
# Ask the user to enter some numbers, separated by spaces.
input_numbers = input("Enter some numbers separated by spaces: ")

# Split the string into parts wherever there is a space.
parts = input_numbers.split()

# Make a new empty list to store numbers.
numbers = []

# Look at every piece from the list of parts.
for p in parts:
    # Turn the string into a number and put it in our list.
    numbers.append(float(p))

# Add up all the numbers using sum(). Get the number of values using len().
average = sum(numbers) / len(numbers)

# Show the user the answer.
print("The average is:", average)
```

**Why and When:**  
- `input()` is used to let users type in data from the keyboard.
- `split()` makes a list of separate strings, making it easy to handle each number.
- `append()` adds each number to the list, one by one.
- This is useful whenever needing to calculate simple statistics—like grades or prices—without external tools.
- Averages help summarize data quickly and simply.

### Project 2: Intermediate Data Analytics – Word Frequency Counter

**Goal:**  
Analyze any block of text to show how often each word appears.

```python
# Ask the user to paste or type a sentence or paragraph.
text = input("Type or paste your text: ")

# Make all letters lowercase so 'Apple' and 'apple' are the same.
text = text.lower()

# Split the text into words using spaces.
words = text.split()

# Make an empty dictionary to store word counts.
word_count = {}

# Look at each word in the list.
for word in words:
    # If the word is already counted, add 1.
    if word in word_count:
        word_count[word] += 1
    # If not, start its counter at 1.
    else:
        word_count[word] = 1

# Show each word and its count.
print("Word frequencies:")
for word, count in word_count.items():
    print(word, ":", count)
```

**Why and When:**  
- Dictionaries help record unique words and counts (like a tally sheet).
- This can be used for reviewing essays, analyzing reviews, or finding common themes.
- `split()` separates text into words for easy counting.
- Shows the basics of data analytics and how to analyze any collection of words quickly.

## Project 2: Intermediate Data Analytics – Word Frequency Counter
Python Only

Analyze any block of text to show how often each word appears.

````python
text = ("Hi I am Mohd Furkhan Hussain, A Billionair with a B")
text = text.lower()

words = text.split()

word_count = {}

for i in words:
    if i in word_count:
        word_count[i] += 1

    else:
        word_count[i] = 1

print("word frequency:")
for i, count in word_count.items():
    print(i, ":", count)
````

## Project 1: Stock Price Moving Average Tracker
only using python

````python
prices = [100, 102, 101, 104, 106, 108]
window_size = 3

moving_average = []

for i in range(len(prices) - window_size + 1):
    window = prices[i:i + window_size]

average = sum(window)/window_size
moving_average.append(average)

print("Moving averages:", moving_average)
````

## 4 portfolio calculator

````python
portfolio = [
    {"name": "Apple", "shares":2, "price":145},
    {"name": "Google", "shares":3, "price":2750},
    {"name": "Infosys", "shares":2, "price":1620}
]

total_value = 0

for stock in portfolio:
    company_value = stock["shares"] * stock["price"]
    total_value += company_value
    print(stock["name"], ":", company_value)

print("Total portfolio value:", total_value)
````

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
features = [[150,0],
             [170,0],
             [130,1],
             [120,1]]
```

```python
labels = ["Apples", "Apples", "Orange", "Orange"]
```

```python
cls = DecisionTreeClassifier()
```

```python
cls = cls.fit(features, labels)
```

```python
prediction = cls.predict([[150,1]])
```

```python
cls.predict([[1510,1], [34, 1]])
```

```python
print(f"The fruite is {prediction[0]}")
```

```python
from sklearn.tree import plot_tree
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(11,6))
```

```python
plot_tree(cls, feature_names=["weight", "color"], class_names=["Apple", "Orange"], filled=True)
```

```python
import json
```

```python
def load_data(filename):
    with open (filename, "r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("Py project Practice 1.json")
```

```python
data
```

```python
def display_users(data):
    print("Users and their connection")
    for user in data['users']:
        print(f"ID:{user['id']} - {user['name']} is friend with:  {user['friends']} and like pages are {user['liked_Games']}")
    print("\nGames Information")
    for page in data['Games']:
        print(f"{page['id']}: {page['name']}")

display_users(data)
```

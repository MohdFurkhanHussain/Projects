```python
from bs4 import BeautifulSoup

with open("htmls/page1.html") as f:
    content = f.read()

soup = BeautifulSoup(content, "html.parser")

h3s = soup.find_all("h3")
```

```python
articles = soup.select("article.product_pod")
```

```python
for article in articles:
    title = article.find("h3").find("a")["title"]
    print(title)
```

```python
for article in articles:
    title = article.find("h3").find("a")["title"]
    price = article.select_one("p.price_color").text
    print(price)
```

```python
for article in articles:
    title = article.find("h3").find("a")["title"]
    price = article.select_one("p.price_color").text.split("£")[1]
    print(price)
```

```python
items = []
for article in articles:
    title = article.find("h3").find("a")["title"]
    price = article.select_one("p.price_color").text.split("£")[1]
    items.append([title, price])
```

```python
items
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(items)
```

```python
df
```

```python
items = []
for article in articles:
    title = article.find("h3").find("a")["title"]
    price = article.select_one("p.price_color").text.split("£")[1]
    rating_elements = article.select_one("p.star-rating")
    rating = rating_elements["class"][1]
    items.append([title, price, rating])
```

```python
df = pd.DataFrame(items, columns = ["Book title", "Price", "Rating"])
```

```python
df
```

```python
df.to_csv("bs4_data.csv",index=False)
```

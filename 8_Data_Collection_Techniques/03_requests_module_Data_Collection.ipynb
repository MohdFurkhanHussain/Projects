```python
import requests
```

```python
a = requests.get("https://books.toscrape.com/")
```

```python
with open(f"html/page.html", "w") as f:
    f.write(a.text)
```

#### next code for download all the pages

```python
for i in range(1, 51):
    a = requests.get(f"https://books.toscrape.com/catalogue/page-{i}.html")
    with open(f"html/page{i}.html", "w", encoding = "utf-8") as f:
        f.write(a.text)
        print(f"Downloaded page (i) successfully")
```

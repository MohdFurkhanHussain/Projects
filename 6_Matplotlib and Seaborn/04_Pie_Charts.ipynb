```python
# Data
labels = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]

#plt.title()
#plt.pie()

plt.title("Career Runs of Indian Batsmen")
plt.pie(runs, labels=labels)
plt.show()
```

```python
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
plt.pie(runs, labels=labels, colors=colors)
plt.show()
```

```python
plt.pie(
    runs,
    labels=labels,
    colors=colors,
    wedgeprops={'edgecolor': 'black', 'linewidth': 2, 'linestyle': '-.'}
)
plt.show()
```

```python
plt.pie(
    runs,
    labels=labels,
    colors=colors,
    wedgeprops={'edgecolor': 'black', 'linewidth': 2, 'linestyle': '-.', 'fill':True }
)
plt.show()
```

```python
plt.pie(
    runs,
    labels=labels,
    colors=colors,
    wedgeprops={'edgecolor': 'black', 'linewidth': 2, 'linestyle': '-.', 'fill':False }
)
plt.show()
```

```python
explode = [0.1, 0, 0, 0]     #only highlight sachins slice
plt.pie(runs, labels=labels, explode=explode, colors=colors)
plt.title("Exploted Pie Example")
```

```python
explode = [0.1, 0.4, 0, 0]     #only highlight sachins slice
plt.pie(runs, labels=labels, explode=explode, colors=colors)
plt.title("Exploted Pie Example")
```

```python
explode = [0.1, 0.4, 0.1, 0.2]     #only highlight sachins slice
plt.pie(runs, labels=labels, explode=explode, colors=colors)
plt.title("Exploted Pie Example")
```

```python
explode = [0.1, 0, 0, 0]     #only highlight sachins slice
plt.pie(runs, labels=labels, explode=explode, colors=colors)
plt.title("Exploted Pie Example")
```

```python
explode = [0.1, 0, 0, 0]     #only highlight sachins slice
plt.pie(runs, labels=labels, explode=explode, colors=colors, shadow = True)
plt.title("Exploted Pie Example")
```

```python
explode = [0.1, 0, 0, 0]     
plt.pie(runs, labels=labels, explode=explode, colors=colors, shadow = True, startangle = 140)
plt.title("Exploted Pie Example")
```

```python
explode = [0.1, 0, 0, 0]     
plt.pie(runs, labels=labels, explode=explode, colors=colors, shadow = True, startangle = 140, autopct = '%1.1f%%')
plt.title("Exploted Pie Example")
```

```python
# Too many categories
languages = ['Python', 'Java', 'C++', 'JavaScript', 'C#', 'Ruby', 'Go', 'Rust', 'Swift', 'PHP']
usage = [30, 20, 10, 10, 7, 5, 4, 3, 2, 1]

plt.pie(usage, labels=languages, autopct='%1.1f%%', startangle=90)
plt.title("Programming Language Usage (Crowded Example)")
plt.show()
```

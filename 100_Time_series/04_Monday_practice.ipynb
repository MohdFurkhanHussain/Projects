```python
import numpy as np
import pandas as pd
import xgboost as XGBRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
# 1. Make some stories of some study and marks
hours = np.array([2, 3, 4, 5, 6, 8, 7, 10, 9, 11])
marks = np.array([50, 55, 60, 68, 75, 85, 80, 95, 88, 99])
df = pd.DataFrame({"hours": hours, "marks": marks})

x = df[["hours"]].values
y = df["marks"].values

# 2 split into learn set and quiz set
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=99)

# 3 Make and teach the regressors
model = XGBRegressor()
model.fit(x_train, y_train)

# 4 Make predictions and see how close we were
pred = model.predict(x_test)
print("True marks:", y_test)
print("predict mark:", np.round(pred))
print("MSE", mean_squared_error(y_test, pred))
```

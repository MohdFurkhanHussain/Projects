```python
# Import the libraries we need
import numpy as np
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.datasets import make_classification

# Step 1: Create a sample dataset (like toy data to practice)
X, y = make_classification(n_samples=100, n_features=2, n_informative=2, 
                           n_redundant=0, n_clusters_per_class=1, 
                           weights=[0.9, 0.1], random_state=42)

# Step 2: Print the class distribution before splitting
unique, counts = np.unique(y, return_counts=True)
print("Original class distribution:")
print(dict(zip(unique, counts)))

# Step 3: Create StratifiedShuffleSplit object
sss = StratifiedShuffleSplit(n_splits=3, test_size=0.3, random_state=42)

# Step 4: Split the data and print results for each split
split_number = 1
for train_index, test_index in sss.split(X, y):
    # Get training and testing data
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = y[train_index], y[test_index]
    
    # Check class distribution in train and test sets
    train_unique, train_counts = np.unique(y_train, return_counts=True)
    test_unique, test_counts = np.unique(y_test, return_counts=True)
    
    print(f"\n--- Split {split_number} ---")
    print(f"Train set size: {len(y_train)}")
    print(f"Train class distribution: {dict(zip(train_unique, train_counts))}")
    print(f"Test set size: {len(y_test)}")
    print(f"Test class distribution: {dict(zip(test_unique, test_counts))}")
    
    split_number += 1
```

```python
import numpy as np         # This code was generated by Perplexcity

arr = np.array([35.0, 75.0, 95.0, 66, 77, 88, 99])

even_mask = arr % 2 == 0       # Boolean mask: True where element is even
even_numbers = arr[even_mask]  # Fancy indexing to get even numbers

print(even_numbers)            # Output: [66. 88.]
```

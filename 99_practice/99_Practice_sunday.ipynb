```python
# Step 2
# Create sample sales data for 4 weeks
weeks = ['week 1', 'week 2', 'week 3', 'week 4']
sales = [12000, 15000, 18000, 21000] #in rupees
```

```python
# Step 3
# Make it a DataFrame
df = pd.DataFrame({'Week': weeks, 'Sales': sales})
```

```python
# step 4
# Calacuate weekly percentage growth
# Formula : ((This week - previous week) / previous week) * 100
df['Growth %'] = df['Sales'].pct_change() * 100
```

```python
# Step 5
# replace the first row's NAN (because it has no previou week) with 0
df ['Growth %'] = df['Growth %'].fillna(0)
```

```python
# Step 6
print (df)
```

```python
# Step 7
#Create two graphs — one for Sales and one for Growth
plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.bar(df['Week'], df['Sales'], color='orange')
plt.title('Weekly Sales')

plt.subplot(1,2,2)
plt.plot(df['Week'], df['Growth %'], color='green', marker='o')
plt.title('Weekly Growth %')
plt.tight_layout()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Step 1: Create fake stock prices for 120 days
np.random.seed(42)
days = np.arange(1, 121)
base_price = 100
noise = np.random.normal(0, 5, 120)
trend = days*0.5 
stock_price = base_price + trend + noise
```

```python
# Step 2: Put the data in a pandas DataFrame
df = pd.DataFrame({
    'Day': days,
    'Price': stock_price
})
```

```python
# Step 3: Calculate 7-day moving average
df['MA_7'] = df['Price'].rolling(window=7).mean()
```

```python
# Step 4: Calculate 30-day moving average
df['MA_30'] = df['Price'].rolling(window=30).mean()
```

```python
# Step 5: Find days when price crossed above the 7-day average
df['Signal'] = np.where(df['Price'] > df['MA_7'], 1, 0)
```

```python
# Step 6: Count how many buy signals we got
buy_signals = df['Signal'].sum()
```

```python
# Step 7: Print results
print('Stock Price Analysis')
print("=" * 50)
print(f"Total Days : {len(df)}")
print(f"starting price: ${df['Price'].iloc[0]:.2f}")
print(f"Ending Price: ${df['Price'].iloc[-1]:.2f}")
print(f"Price Change: ${df['Price'].iloc[-1] - df['Price'].iloc[0]:.2f}")
print(f"Buy Signals Generated: {buy_signals}")
print("\nFirst 10 Days:")
print(df.head(10))
```

```python
import numpy as np
import pandas as pd
from collections import Counter
```

```python
# Step 1: Define positive and negative words
Positive_Words = ['good', 'great', 'excellent', 'amazing', 'wonderful', 
                  'fantastic', 'love', 'best', 'happy', 'perfect']

Negative_Words = ['bad', 'terrible', 'awful', 'horrible', 'worst', 
                  'hate', 'poor', 'disappointing', 'sad', 'angry']
```

```python
# Step 2: Sample customer reviews
reviews = ["This product is great and wonderful! I love it.",
    "Terrible experience. Worst purchase ever. Very disappointing.",
    "Good quality but poor customer service. Not happy.",
    "Amazing product! Best thing I ever bought. Fantastic!",
    "Bad quality. Awful design. Hate it.",
    "Excellent service and perfect product. Very happy!",]
```

```python
# Step 3: Function to analyze sentiment
def analyze_sentiment(text):
    # Convert to lowercase and split into words
    words = text.lower().split()

    # Count positive and negative words
    pos_count = sum(1 for word in words if word.strip('.,!?') in positive_words )
    neg_count = sum(1 for word in words if word.strip('.,!?') in negative_words )

    # Calculate sentiment score
    sentiment_score = pos_count - neg_count

    # Determine overall sentiment
    if sentiment_score > 0:
        sentiment = 'Positive'
    elif sentiment_score < 0:
        sentiment = 'Negative'
    else:
        sentiment = 'Neutral'

    return pos_count, neg_count, sentiment_score, sentiment
```

```python
# Step 4: Analyze all reviews
results = []
for i, review in enumerate(reviews, 1):
    pos, neg, score, sentiment = analyze_sentiment(review)
    results.append({
        'Review_Number': i,
        'Review': review,
        'Positive_Words': pos,
        'Negative_Words': neg,
        'Sentiment_Scores': score,
        'Sentiment': sentiment, 
    })
```

```python
# Step 5: Create DataFrame
df = pd.DataFrame(results)
```

```python
# Step 6: Calculate statistics
total_positive = df['Positive_Words'].sum()
total_negative = df['Negative_Words'].sum()
avg_score = df['Sentiment_Scores'].mean()
```

```python
# Step 7: Print results
print("SENTIMENT ANALYSIS REPORT")
print("=" * 60)
print(f"Total Review Analyzed: {len(df)}")
print(f"Total Positive Words Found: {total_positive}")
print(f"Total Negative words Found: {total_negative}")
print(f"Average Sentiment Score: {avg_score:.2f}")
print("\nDetailed Results:")
print(df.to_string(index=False))
print("\nSentiment Distributation:")
print(df["Sentiment"].value_counts())
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Generate values
x = np.linspace(-4, 4, 1000)
mean = 0
std_dev = 1

# Get the probability density
y = norm.pdf(x, loc=mean, scale=std_dev)

# Plot
plt.plot(x, y)
plt.title("Standard Normal Distribution (μ=0, σ=1)")
plt.xlabel("x")
plt.ylabel("Probability Density")
plt.grid(True)
plt.show() 
```
